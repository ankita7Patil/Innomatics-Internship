```python
from langchain_core.prompts import PromptTemplate
from langchain_core.runnables import RunnableLambda

def fake_llm(prompt):
    return "LangChain is a framework for building LLM applications."

llm = RunnableLambda(fake_llm)

prompt = PromptTemplate.from_template(
    "Explain {topic}"
)

chain = prompt | llm

response = chain.invoke({"topic":"LangChain"})
print(response)
```

```python
chat_history = []

def memory_llm(input_text):
    chat_history.append(input_text)
    return f"Previous messages: {chat_history}"

memory_chain = RunnableLambda(memory_llm)

print(memory_chain.invoke("Hi"))
print(memory_chain.invoke("What is LangChain?"))
```

```python
def calculator(x):
    return x * 10

print(calculator(5))
```

```python
def agent(question):
    if "multiply" in question:
        return "Using calculator tool: " + str(calculator(5))
    else:
        return "Using LLM: LangChain helps build AI apps"

print(agent("multiply number"))
```
