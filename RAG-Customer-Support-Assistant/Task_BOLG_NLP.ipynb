```python
# NLP Preprocessing Engine

import re
import string
import nltk
nltk.download('stopwords')
nltk.download('wordnet')
nltk.download('punkt')
nltk.download('punkt_tab') # Added to resolve LookupError

from nltk.corpus import stopwords
from nltk.stem import PorterStemmer, WordNetLemmatizer
from nltk.tokenize import word_tokenize

# Initialize tools
stop_words = set(stopwords.words('english'))
stemmer = PorterStemmer()
lemmatizer = WordNetLemmatizer()

# Sample text
text = "Hello!!! 😊 This is an NLP task. Visit https://example.com for more info!!! #AI #NLP"

# 1. Lowercasing
def to_lower(text):
    return text.lower()

# 2. Remove URLs
def remove_urls(text):
    return re.sub(r'http\S+|www\S+', '', text)

# 3. Remove Emojis
def remove_emojis(text):
    return re.sub(r'[^\w\s,]', '', text)

# 4. Remove punctuation
def remove_punctuation(text):
    return text.translate(str.maketrans('', '', string.punctuation))

# 5. Tokenization
def tokenize(text):
    return word_tokenize(text)

# 6. Remove Stopwords
def remove_stopwords(tokens):
    return [word for word in tokens if word not in stop_words]

# 7. Stemming
def stemming(tokens):
    return [stemmer.stem(word) for word in tokens]

# 8. Lemmatization
def lemmatization(tokens):
    return [lemmatizer.lemmatize(word) for word in tokens]

# Full pipeline
def preprocess(text):
    text = to_lower(text)
    text = remove_urls(text)
    text = remove_emojis(text)
    text = remove_punctuation(text)
    tokens = tokenize(text)
    tokens = remove_stopwords(tokens)
    stemmed = stemming(tokens)
    lemmatized = lemmatization(tokens)

    return {
        "clean_text": text,
        "tokens": tokens,
        "stemmed": stemmed,
        "lemmatized": lemmatized
    }

# Run
output = preprocess(text)

print("Clean Text:", output["clean_text"])
print("Tokens:", output["tokens"])
print("Stemmed:", output["stemmed"])
print("Lemmatized:", output["lemmatized"])
```
