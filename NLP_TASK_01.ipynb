# Task 1: Conceptual Understanding

1.Difference between "Love" and "love" in NLP

In Natural Language Processing, "Love" and "love" are treated as two different words if the text is case-sensitive. This increases the number of unique words (vocabulary size) unnecessarily. To avoid this issue, text is usually converted into lowercase so that both "Love" and "love" are treated as the same word.


2.What happens if stopwords are not removed?

If stopwords (like "is", "the", "and") are not removed:

The dataset becomes noisy
Model training becomes slower
Important words may get less importance
Accuracy of the model can decrease


3.Two real-world scenarios where removing stopwords can be harmful
Sentiment Analysis
Example: "I am not happy"
If "not" is removed, the meaning becomes "I am happy", which is incorrect.
Question Answering Systems
Example: "What is your name?"
Removing words like "what" and "is" can make the sentence meaningless.


4.Difference between Stemming and Lemmatization
Stemming:
It reduces words to their root form by cutting parts of the word.
Example: "running" → "runn"
Lemmatization:
It converts words to their meaningful base form using vocabulary and grammar.
Example: "running" → "run"

# Task 2: Build Advanced Preprocessing Function

```python
import re

def preprocess_text(text):

    if not isinstance(text, str) or text.strip() == "":
        return [], ""

    text = text.lower()

    text = re.sub(r'http\S+|www\S+|\S+@\S+', '', text)

    text = re.sub(r'\d+', '', text)


    text = re.sub(r'[^\w\s]', ' ', text)

    text = re.sub(r'(.)\1{2,}', r'\1\1', text)


    text = re.sub(r'\s+', ' ', text).strip()


    tokens = text.split()

    tokens = [word for word in tokens if len(word) > 2 or word in ["no", "not"]]

    clean_sentence = " ".join(tokens)

    return tokens, clean_sentence
```

```python
# Test Sentences
sentences = [
    "Get 100% FREE access now!!!",
    "I absolutely looooved this product 😍😍",
    "Worst service ever... 0/10",
    "Call me at 9876543210",
    "This is THE best course!!!",
    "Visit https://openai.com now!",
    "Nooooo this is baaad!!!",
    "OK OK OK I got it",
    "Win $$$ now!!! Limited offer!!!",
    "I am not happy with this"
]

print("=== STRESS TESTING OUTPUT ===\n")

for text in sentences:
    tokens, clean = preprocess_text(text)

    print("Original Text:", text)
    print("Cleaned Tokens:", tokens)
    print("Cleaned Sentence:", clean)
    print("-" * 50)
```

# Task 4: Token Analytics


```python
print("\n=== TOKEN ANALYTICS ===\n")

results = []

for text in sentences:
    tokens, clean = preprocess_text(text)

    total_tokens = len(tokens)
    unique_tokens = len(set(tokens))

    avg_length = 0
    if total_tokens > 0:
        avg_length = round(sum(len(word) for word in tokens) / total_tokens, 2)

    results.append({
        "original": text,
        "tokens": tokens,
        "total": total_tokens
    })

    print("Sentence:", text)
    print("Total Tokens:", total_tokens)
    print("Unique Tokens:", unique_tokens)
    print("Average Token Length:", avg_length)
    print("-" * 50)


# Analysis Answers
most_noise = max(results, key=lambda x: len(x["original"]))
most_meaningful = max(results, key=lambda x: x["total"])

print("\n=== ANALYSIS ANSWERS ===")
print("Most noisy sentence:", most_noise["original"])
print("Most meaningful sentence:", most_meaningful["original"])
```

# Task 5: Frequency Analysis

```python
from collections import Counter

print("\n=== FREQUENCY ANALYSIS ===\n")

all_tokens = []

# Collect all tokens
for text in sentences:
    tokens, _ = preprocess_text(text)
    all_tokens.extend(tokens)

# Count frequency
counter = Counter(all_tokens)

# Top 10 frequent words
print("Top 10 Most Frequent Words:")
print(counter.most_common(10))

# Top 5 least frequent words
print("\nTop 5 Least Frequent Words:")
print(counter.most_common()[:-6:-1])
```

# Task 6: Build Full Pipeline

```python
def full_pipeline(text_list):
    all_tokens = []
    clean_sentences = []

    for text in text_list:
        tokens, clean = preprocess_text(text)
        all_tokens.extend(tokens)
        clean_sentences.append(clean)

    return {
        "tokens": all_tokens,
        "clean_sentences": clean_sentences
    }


# Run pipeline
pipeline_output = full_pipeline(sentences)

print("\n=== FULL PIPELINE OUTPUT ===\n")
print("All Tokens:\n", pipeline_output["tokens"])
print("\nClean Sentences:\n", pipeline_output["clean_sentences"])
```
